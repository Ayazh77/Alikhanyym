```python
def count_iqr_outliers(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    return ((series < lower) | (series > upper)).sum()


numeric_columns = df.select_dtypes(include="number").columns

outlier_counts = {
    column: count_iqr_outliers(df[column].dropna())
    for column in numeric_columns
}

print("Количество выбросов:")
print(pd.Series(outlier_counts).sort_values(ascending=False))
```

```python
X = df.drop(columns="median_house_value")
y = df["median_house_value"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=21
)

print("Обучающая выборка:", X_train.shape)
print("Тестовая выборка:", X_test.shape)
```

```python
numeric_features = X_train.select_dtypes(include="number").columns.tolist()
categorical_features = X_train.select_dtypes(exclude="number").columns.tolist()

print("Числовые признаки:")
print(numeric_features)

print("\nКатегориальные признаки:")
print(categorical_features)
```

```python
class IQRClipper(BaseEstimator, TransformerMixin):
    def __init__(self, factor=1.5):
        self.factor = factor

    def fit(self, X, y=None):
        X = np.asarray(X, dtype=float)

        q1 = np.nanpercentile(X, 25, axis=0)
        q3 = np.nanpercentile(X, 75, axis=0)

        iqr = q3 - q1

        self.lower_bounds_ = q1 - self.factor * iqr
        self.upper_bounds_ = q3 + self.factor * iqr

        return self

    def transform(self, X):
        X = np.asarray(X, dtype=float)

        return np.clip(
            X,
            self.lower_bounds_,
            self.upper_bounds_
        )

    def get_feature_names_out(self, input_features=None):
        return np.asarray(input_features, dtype=object)
```

```python
numeric_pipeline = Pipeline(steps=[
    ("outliers", IQRClipper(factor=1.5)),
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

numeric_pipeline
```

```python
categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

categorical_pipeline
```

```python
preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

preprocessor
```

```python
model_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])

model_pipeline
```

```python
model_pipeline.fit(X_train, y_train)
```

```python
y_pred = model_pipeline.predict(X_test)

rmse = mean_squared_error(y_test, y_pred) ** 0.5
r2 = r2_score(y_test, y_pred)

rmse, r2
```

```python
minmax_pipeline = Pipeline(steps=[
    ("outliers", IQRClipper(factor=1.5)),
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", MinMaxScaler())
])

preprocessor_minmax = ColumnTransformer(transformers=[
    ("num", minmax_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

model_pipeline_minmax = Pipeline(steps=[
    ("preprocessor", preprocessor_minmax),
    ("model", LinearRegression())
])

model_pipeline_minmax.fit(X_train, y_train)

y_pred_minmax = model_pipeline_minmax.predict(X_test)

rmse_minmax = mean_squared_error(y_test, y_pred_minmax) ** 0.5
r2_minmax = r2_score(y_test, y_pred_minmax)

rmse_minmax, r2_minmax
```

```python
def count_clipped_values(X, factor):
    X = X[numeric_features].copy()

    q1 = X.quantile(0.25)
    q3 = X.quantile(0.75)
    iqr = q3 - q1

    lower = q1 - factor * iqr
    upper = q3 + factor * iqr

    clipped = ((X < lower) | (X > upper)).sum()

    return clipped.sum()

clipped_15 = count_clipped_values(X_train, 1.5)
clipped_30 = count_clipped_values(X_train, 3.0)

clipped_15, clipped_30
```
